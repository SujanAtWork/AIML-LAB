```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = "dataset/Environment_Temperature_change_E_All_Data_NOFLAG.csv"

df = pd.read_csv(
    file_path,
    encoding="latin1"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.tail()
```

## Load the dataset

```python
print("Columns in dataset:")
for i, column in enumerate(df.columns):
    print(i, ":", column)
```

```python
df.info()
```

```python
print(df["Element"].unique())
```

```python
india = df[df["Area"] == "India"].copy()

print("India records:", india.shape)

india.head()
display(india)
```

## Calulate errors


```python
india_temp = df[
    (df["Area"] == "India") &
    (df["Element"] == "Temperature change")
].copy()

print("Shape:", india_temp.shape)

india_temp[["Area", "Months", "Element", "Unit"]].head(15)
```

## Keep only january to december

```python
month_names = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

india_monthly = india_temp[
    india_temp["Months"].isin(month_names)
].copy()

print("Number of monthly records:", len(india_monthly))

display(
    india_monthly[["Months", "Element", "Unit"]]
)
```

```python
year_columns = [
    col for col in df.columns
    if col.startswith("Y")
]

print("Number of years:", len(year_columns))
print(year_columns[0], "to", year_columns[-1])
```

# Calculate Average temperature for each month

```python
india_monthly["Average_Temperature_Change"] = (
    india_monthly[year_columns]
    .apply(pd.to_numeric, errors="coerce")
    .mean(axis=1)
)

monthly_temperature = india_monthly[
    ["Months", "Average_Temperature_Change"]
].copy()

monthly_temperature
```

```python
month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

monthly_temperature["Month"] = (
    monthly_temperature["Months"].map(month_mapping)
)

monthly_temperature = monthly_temperature.sort_values(
    "Month"
).reset_index(drop=True)

display(monthly_temperature)
```

### Define x and y

```python
X = monthly_temperature[["Month"]]

y = monthly_temperature["Average_Temperature_Change"]

print("X (Independent Variable):")
display(X)

print("\ny (Dependent Variable):")
display(y)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=0
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
intercept = linear_model.intercept_
coefficient = linear_model.coef_[0]

print("Intercept (a0):", intercept)
print("Coefficient (a1):", coefficient)

print(
    f"\nRegression Equation:"
    f"\nTemperature Change = {intercept:.4f} "
    f"+ ({coefficient:.4f} × Month)"
)
```

```python
y_pred = linear_model.predict(X_test)

results = pd.DataFrame({
    "Month": X_test["Month"].values,
    "Actual Temperature Change": y_test.values,
    "Predicted Temperature Change": y_pred
})

results = results.sort_values("Month")

display(results)
```

```python
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Linear Regression Performance")
print("--------------------------------")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-Square (R²): {r2:.4f}")
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    X,
    y,
    label="Actual Temperature Change"
)

X_line = np.linspace(
    1,
    12,
    100
).reshape(-1, 1)

y_line = linear_model.predict(X_line)

plt.plot(
    X_line,
    y_line,
    label="Regression Line"
)

plt.title(
    "India Month-wise Average Temperature Change"
)

plt.xlabel("Month")
plt.ylabel("Temperature Change (°C)")

plt.xticks(
    range(1, 13),
    month_names,
    rotation=45
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_temperature["Months"],
    monthly_temperature["Average_Temperature_Change"],
    marker="o"
)

plt.title(
    "India Month-wise Average Temperature Change (1961–2019)"
)

plt.xlabel("Month")
plt.ylabel("Average Temperature Change (°C)")

plt.xticks(rotation=45)

plt.grid(True)
plt.tight_layout()

plt.show()
```
